## 可視化する例を実施

```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
df = pd.DataFrame({"month":[1,2,3,4,5], "total_sales":[10, 20, 30, 40, 50], "previous_total_sales":[20, 40, 60, 10, 10],})
fig = plt.figure(figsize=[4.5, 3.7])
fig.suptitle('Sales Graph')

ax = fig.add_subplot()
ax.plot(df['month'], df['total_sales'], label='this year')
ax.plot(df['month'], df['previous_total_sales'], label='previous')
ax.set_xlabel('month')
ax.set_ylabel('total_sales')

plt.legend(loc='upper right')
plt.show()
```
